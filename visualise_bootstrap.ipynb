# Visualisation scripts for observation-only augmentations \& Bootstrap with additional transformations

# 1. Gaussian experiments

## 1.1 Vary with $\alpha$

```python
import matplotlib.pyplot as plt
from matplotlib.legend_handler import HandlerTuple
import matplotlib as mpl
from utils.plotting import CISeries, plot_CI_trend_fn, plot_CI_freq_fn

k = 500
d = 2
small_n = 5
large_n = 30

folder=f'logs/bootstrap/bootstrap2dGaussian_n{small_n}-{large_n}_d{d}_k{k}_alpha0.00-0.30_seed0_sim500'

small_series = [
    CISeries(
        label=f'{label}, n={small_n}',
        paths=[f'{folder}/_n{small_n}_k{k}.npz'],
        varfield='alpha',
        CIfield=CIfield,
        color=color,
        linestyle="-o",
        hist_linestyle="-",
        linewidth=1,
        markersize=5,
    )
    for label, CIfield, color in zip(
        ['Gaussian', 'Ortho', 'Bootstrap', 'Bootstrap + Ortho'], 
        ['gaussian_cis', 'random_ortho_cis', 'bootstrap_compose_id_cis', 'bootstrap_compose_random_ortho_cis', 'bootstrap_average_random_ortho_cis'],
        ['black', 'olive', 'tab:blue', 'tab:red', 'tab:purple']
    )
]


large_series = [
    CISeries(
        label=f'{label}, n={large_n}',
        paths=[f'{folder}/_n{large_n}_k{k}.npz'],
        varfield='alpha',
        CIfield=CIfield,
        color=color,
        linestyle=":^",
        hist_linestyle=":",
        linewidth=1,
        markersize=5,
    )
    for label, CIfield, color in zip(
        ['Gaussian', 'Ortho', 'Bootstrap', 'Bootstrap + Ortho'], 
        ['gaussian_cis', 'random_ortho_cis', 'bootstrap_compose_id_cis', 'bootstrap_compose_random_ortho_cis', 'bootstrap_average_random_ortho_cis'],
        ['black', 'olive', 'tab:blue', 'tab:red', 'tab:purple']
    )
]

fig, axs = plt.subplots(1, 3, figsize=(15, 5), gridspec_kw={"width_ratios": [1.1, 1.1, 0.8]})

plot_CI_trend_fn(
    fn=lambda l,m,u: l <= 0. and 0. <= u,
    series=small_series + large_series,
    guides=[lambda x:x],
    xlim=[0.7,1.0],
    ylim=[0.55,1.0],
    fontsize=12,
    ax=axs[0],
)

plot_CI_trend_fn(
    fn=lambda l,m,u: u-l,
    series=small_series + large_series,
    xlim=[0.7,1.0],
    ylim=[0.3, 3],
    fontsize=12,
    ax=axs[1],
)

plot_CI_freq_fn(
    var_val=0.95,
    fn=lambda l,u,m: u-l,
    series=small_series + large_series,
    num_bins=30,
    xlim=[0.1,2.5],
    ylim=[0,3],
    transparency=0.4,
    ax=axs[2],
)

# pick up legends from axs[1] and axs[2]
legend_by_label = {}
for ax in (axs[1], axs[2]):
    handles, labels = ax.get_legend_handles_labels()
    for handle, label in zip(handles, labels):
        legend_by_label.setdefault(label, []).append(handle)

legend_handles = []
legend_labels = []
for label, handles in legend_by_label.items():
    if len(handles) == 1:
        legend_handles.append(handles[0])
    else:
        legend_handles.append(tuple(handles))
    legend_labels.append(label)

# reorder 
new_order = [0,4,1,5,2,6,3,7] 
legend_handles = [legend_handles[i] for i in new_order]
legend_labels = [legend_labels[i] for i in new_order]

# Leave space below axs[0] for the legend and place it at the bottom-left of the overview.
fig.tight_layout()
fig.subplots_adjust(wspace=0.2)
pos0 = axs[0].get_position()
shrink = pos0.height * 0.25
axs[0].set_position([pos0.x0, pos0.y0, pos0.width, pos0.height - shrink])
pos1 = axs[1].get_position()
shrink = pos1.height * 0.25
axs[1].set_position([pos1.x0, pos1.y0, pos1.width, pos1.height - shrink])
pos2 = axs[2].get_position()
shrink = pos2.height * 0.25
axs[2].set_position([pos2.x0, pos2.y0, pos2.width, pos2.height - shrink])


for ax in axs:
    ax.tick_params(axis='both', labelsize=14)
    

fig.legend(
    legend_handles,
    legend_labels,
    loc='upper left',
    bbox_to_anchor=(pos0.x0+0.05, pos0.y0 + pos0.height + 0.045),
    bbox_transform=fig.transFigure,
    handlelength=4.0,
    handler_map={tuple: HandlerTuple(ndivide=None)},
    ncol=4,
    fontsize=14.4,
)
fig.savefig('figs/bootstrap2dgaussian_overview.pdf', dpi=100, bbox_inches='tight')
```

## 1.2 Vary with $n$ and $k$

```python
import matplotlib.pyplot as plt
from matplotlib.legend_handler import HandlerTuple
import matplotlib as mpl
from utils.plotting import CISeries, plot_CI_trend_fn, plot_CI_freq_fn
import numpy as np

alpha = 0.05

k = 500
d = 2

folder_n=f'logs/bootstrap/bootstrap2dGaussian_n5-30_d{d}_k{k}_alpha{alpha}_seed0_sim500'
n_series = [
    CISeries(
        label=f'{label}',
        paths=[f'{folder_n}/_n{n}_k{k}.npz' for n in range(5,31)],
        varfield='n',
        alpha=alpha,
        CIfield=CIfield,
        color=color,
        linestyle="-o",
        hist_linestyle="-",
        linewidth=1,
        markersize=5,
    )
    for label, CIfield, color in zip(
        ['Gaussian', 'Ortho', 'Bootstrap', 'Bootstrap + Ortho'], 
        ['gaussian_cis', 'random_ortho_cis', 'bootstrap_compose_id_cis', 'bootstrap_compose_random_ortho_cis', 'bootstrap_average_random_ortho_cis'],
        ['black', 'olive', 'tab:blue', 'tab:red', 'tab:purple']
    )
]


n = 5
folder_k=f'logs/bootstrap/bootstrap2dGaussian_n{n}_d{d}_k10-485_alpha{alpha}_seed0_sim500'
k_series = [
    CISeries(
        label=f'{label}',
        paths=[f'{folder_k}/_n{n}_k{k}.npz' for k in range(10,501,25)],
        varfield='k',
        alpha=alpha,
        CIfield=CIfield,
        color=color,
        linestyle="-o",
        hist_linestyle="-",
        linewidth=1,
        markersize=5,
    )
    for label, CIfield, color in zip(
        ['Gaussian', 'Ortho', 'Bootstrap', 'Bootstrap + Ortho'], 
        ['gaussian_cis', 'random_ortho_cis', 'bootstrap_compose_id_cis', 'bootstrap_compose_random_ortho_cis', 'bootstrap_average_random_ortho_cis'],
        ['black', 'olive', 'tab:blue', 'tab:red', 'tab:purple']
    )
]

# process Gaussian CI which does not vary with k
count = len(k_series[0].vars)
k_series[0].lowers = k_series[0].lowers.reshape((1,-1)).repeat(count, axis=0)
k_series[0].means = k_series[0].means.reshape((1,-1)).repeat(count, axis=0)
k_series[0].uppers = k_series[0].uppers.reshape((1,-1)).repeat(count, axis=0)


fig, axs = plt.subplots(2, 2, figsize=(12, 6))

plot_CI_trend_fn(
    fn=lambda l,m,u: l <= 0. and 0. <= u,
    series=n_series,
    xlim=[4.9,30.1],
    ylim=[0.7,1.0],
    guides=[lambda x: np.ones_like(x) * 0.95],
    fontsize=12,
    ax=axs[0][0],
)
axs[0][0].xaxis.set(visible=False)

plot_CI_trend_fn(
    fn=lambda l,m,u: u-l,
    series=n_series,
    xlim=[4.9,30.1],
    ylim=[0.8, 2.5],
    fontsize=12,
    ax=axs[1][0],
)

plot_CI_trend_fn(
    fn=lambda l,m,u: l <= 0. and 0. <= u,
    series=k_series,
    xlim=[9.9,485.1],
    ylim=[0.7,1.0],
    guides=[lambda x: np.ones_like(x) * 0.95],
    fontsize=12,
    ax=axs[0][1],
)
axs[0][1].xaxis.set(visible=False)
axs[0][1].yaxis.set(visible=False)

plot_CI_trend_fn(
    fn=lambda l,m,u: u-l,
    series=k_series,
    xlim=[9.9,485.1],
    ylim=[0.8, 2.5],
    fontsize=12,
    ax=axs[1][1],
)
axs[1][1].yaxis.set(visible=False)

# pick up legends from axs[1] 
legend_by_label = {}

handles, labels = axs[0][0].get_legend_handles_labels()
for handle, label in zip(handles, labels):
    legend_by_label.setdefault(label, []).append(handle)

legend_handles = []
legend_labels = []
for label, handles in legend_by_label.items():
    if len(handles) == 1:
        legend_handles.append(handles[0])
    else:
        legend_handles.append(tuple(handles))
    legend_labels.append(label)

# # reorder 
new_order = [0,1,2,3]
legend_handles = [legend_handles[i] for i in new_order]
legend_labels = [legend_labels[i] for i in new_order]

# Leave space below axs[0] for the legend and place it at the bottom-left of the overview.
fig.tight_layout()
fig.subplots_adjust(wspace=0.2)

pos0 = axs[0][0].get_position()
shrink = pos0.height * 0.25
axs[0][0].set_position([pos0.x0, pos0.y0, pos0.width, pos0.height - shrink])
pos1 = axs[0][1].get_position()
shrink = pos1.height * 0.25
axs[0][1].set_position([pos1.x0, pos1.y0, pos1.width, pos1.height - shrink])

pos10 = axs[1][0].get_position()
shrink = pos10.height * 0.25
axs[1][0].set_position([pos10.x0, pos10.y0 + shrink, pos10.width, pos10.height - shrink])
pos11 = axs[1][1].get_position()
shrink = pos11.height * 0.25
axs[1][1].set_position([pos11.x0, pos11.y0 + shrink, pos11.width, pos11.height - shrink])

for ax_row in axs:
    for ax in ax_row:
        ax.tick_params(axis='both', labelsize=14)


fig.legend(
    legend_handles,
    legend_labels,
    loc='upper left',
    bbox_to_anchor=(pos0.x0+0.03, pos0.y0 + pos0.height + 0.07),
    bbox_transform=fig.transFigure,
    handlelength=4.0,
    handler_map={tuple: HandlerTuple(ndivide=None)},
    ncol=5,
    fontsize=15.5,
)

fig.savefig('figs/bootstrap2dgaussian_overview_vary_nk.pdf', dpi=100, bbox_inches='tight')
```
