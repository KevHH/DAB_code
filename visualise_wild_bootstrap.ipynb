### 1.1 Gaussian, vary with $\alpha$

```python
import matplotlib.pyplot as plt
from matplotlib.legend_handler import HandlerTuple
import matplotlib as mpl
from utils.plotting import CISeries, plot_CI_trend_fn, plot_CI_freq_fn
import scipy
import numpy as np

d = 2
k = 500
small_n = 5
large_n = 30

folder=f'logs/WB2dGaussian_n{small_n}-{large_n}_d{d}_k{k}_alpha0.00-0.30_testparam0_seed0_sim500'

small_series = [
    CISeries(
        label=f'{label}, n={small_n}',
        paths=[f'{folder}/_n{small_n}_k{k}.npz'],
        varfield='alpha',
        CIfield=CIfield,
        color=color,
        linestyle="-o",
        hist_linestyle="-",
        linewidth=1,
        markersize=5,
    )
    for label, CIfield, color in zip(
        ['Wild bootstrap', 'WB + Ortho', 'WB + Permute'], 
        ['WB_compose_id_cis', 'WB_compose_random_ortho_cis', 'WB_compose_permute_coord_cis'],
        ['tab:blue', 'tab:red', 'tab:green']
    )
]

large_series = [
    CISeries(
        label=f'{label}, n={large_n}',
        paths=[f'{folder}/_n{large_n}_k{k}.npz'],
        varfield='alpha',
        CIfield=CIfield,
        color=color,
        linestyle=":^",
        hist_linestyle=":",
        linewidth=1,
        markersize=5,
    )
    for label, CIfield, color in zip(
        ['Wild bootstrap', 'WB + Ortho', 'WB + Permute'], 
        ['WB_compose_id_cis', 'WB_compose_random_ortho_cis', 'WB_compose_permute_coord_cis'],
        ['tab:blue', 'tab:red', 'tab:green']
    )
]

fig, axs = plt.subplots(1, 2, figsize=(10, 5))

_, vals = plot_CI_trend_fn(
    fn=lambda l,m,u: l <= 0. and 0. <= u,
    series=small_series + large_series, 
    guides=[lambda x:x],
    xlim=[0.7,1.0],
    # ylim=[0.55,1.0],
    fontsize=12,
    ax=axs[0],
)


plot_CI_trend_fn(
    fn=lambda l,m,u: u-l,
    series=small_series + large_series, 
    xlim=[0.7,1.0],
    # ylim=[1, 4],
    fontsize=12,
    ax=axs[1],
)


# pick up legends from axs[1] and axs[2]
legend_by_label = {}
ax = axs[0]
handles, labels = ax.get_legend_handles_labels()
for handle, label in zip(handles, labels):
    legend_by_label.setdefault(label, []).append(handle)

legend_handles = []
legend_labels = []
for label, handles in legend_by_label.items():
    if len(handles) == 1:
        legend_handles.append(handles[0])
    else:
        legend_handles.append(tuple(handles))
    legend_labels.append(label)

# reorder 
# new_order = [0,1,2,3,4,5] # [0,2,4,1,3,5]
# legend_handles = [legend_handles[i] for i in new_order]
# legend_labels = [legend_labels[i] for i in new_order]


# Leave space below axs[0] for the legend and place it at the bottom-left of the overview.
fig.tight_layout()
fig.subplots_adjust(wspace=0.2)
pos0 = axs[0].get_position()
shrink = pos0.height * 0.05
axs[0].set_position([pos0.x0, pos0.y0, pos0.width, pos0.height - shrink])

pos1 = axs[1].get_position()
shrink = pos1.height * 0.25
axs[1].set_position([pos1.x0, pos1.y0, pos1.width, pos1.height - shrink])

for ax in axs:
    ax.tick_params(axis='both', labelsize=14)
    
fig.legend(
    legend_handles,
    legend_labels,
    loc='upper left',
    bbox_to_anchor=(pos1.x0-0.065, pos1.y0 + pos1.height + 0.05),
    bbox_transform=fig.transFigure,
    handlelength=2.0,
    handler_map={tuple: HandlerTuple(ndivide=None)},
    ncol=2,
    fontsize=12,
)
fig.savefig('figs/wb2dgaussian_overview.pdf', bbox_inches='tight', dpi=100)
```

```python
from pathlib import Path

import jax
import jax.numpy as jnp
import matplotlib.pyplot as plt
import numpy as np
from skimage.transform import resize

from utils.augs import _make_random_rotate, _make_random_rotate_zoom, _make_random_zoom


folder = "data/mnist"
image_train = np.load(f"{folder}/image_train.npy") / 255.0
image_test = np.load(f"{folder}/image_test.npy") / 255.0

MNIST_DOWNSIZE = 14

image_total = np.array(
    [
        resize(img, (MNIST_DOWNSIZE, MNIST_DOWNSIZE), anti_aliasing=True)
        for img in np.concatenate([image_train, image_test], axis=0)
    ]
)

RNG = np.random.default_rng(0)
OUTPUT_DIR = Path("figs")
OUTPUT_DIR.mkdir(exist_ok=True)


def sample_image(images: np.ndarray) -> np.ndarray:
    index = RNG.integers(images.shape[0])
    return images[index]


def add_noise(image: np.ndarray, noise_level: float) -> np.ndarray:
    noisy = image + RNG.normal(size=image.shape) * noise_level
    return np.clip(noisy, 0.0, 1.0)


def apply_transform(
    image: np.ndarray,
    transform,
    key: jax.Array,
) -> np.ndarray:
    transformed = transform(jnp.asarray(image.reshape(-1)), key)
    return np.clip(np.asarray(transformed).reshape(image.shape), 0.0, 1.0)


def save_row(images: list[np.ndarray], titles: list[str], output_path: Path) -> None:
    fig, axes = plt.subplots(1, len(images), figsize=(2.2 * len(images), 2.6))
    for ax, image, title in zip(axes, images, titles):
        ax.imshow(image, cmap="gray", vmin=0.0, vmax=1.0)
        # ax.set_title(title, fontsize=10)
        ax.axis("off")
    fig.tight_layout()
    fig.savefig(output_path, dpi=200, bbox_inches="tight")
    plt.close(fig)


sample = sample_image(image_total)
noise_005 = add_noise(sample, 0.05)
noise_012 = add_noise(sample, 0.12)
noise_020 = add_noise(sample, 0.20)

random_rotate = _make_random_rotate()
random_zoom = _make_random_zoom()
random_rotate_zoom = _make_random_rotate_zoom()


save_row(
    [
        sample,
        noise_005,
        noise_012,
        noise_020,
        apply_transform(noise_012, random_rotate, jax.random.PRNGKey(0)),
        apply_transform(noise_012, random_zoom, jax.random.PRNGKey(0)),
        apply_transform(noise_012, random_rotate_zoom, jax.random.PRNGKey(0)),
    ],
    [
        "original",
        "noise 0.05",
        "noise 0.12",
        "noise 0.20",
        "noise 0.12 + rotate",
        "noise 0.12 + zoom",
        "noise 0.12 + rotate+zoom",
    ],
    OUTPUT_DIR / "mnist_visualisation.pdf",
)
```

```python
from pathlib import Path

import jax
import jax.numpy as jnp
import matplotlib.pyplot as plt
import numpy as np
from skimage.transform import resize

from utils.augs import _make_random_rotate, _make_random_rotate_zoom, _make_random_zoom


folder = "data/cifar10"
image_train = np.load(f"{folder}/image_train.npy") / 255.0
image_test = np.load(f"{folder}/image_test.npy") / 255.0

CIFAR_DOWNSIZE = 16

image_total = np.array(
    [
        resize(img, (CIFAR_DOWNSIZE, CIFAR_DOWNSIZE), anti_aliasing=True)
        for img in np.concatenate([image_train, image_test], axis=0)
    ]
)

RNG = np.random.default_rng(0)
OUTPUT_DIR = Path("figs")
OUTPUT_DIR.mkdir(exist_ok=True)


def sample_image(images: np.ndarray) -> np.ndarray:
    index = RNG.integers(images.shape[0])
    return images[index]


def apply_transform(
    image: np.ndarray,
    transform,
    key: jax.Array,
) -> np.ndarray:
    transformed = transform(jnp.asarray(image.reshape(-1)), key)
    return np.clip(np.asarray(transformed).reshape(image.shape), 0.0, 1.0)


def save_row(images: list[np.ndarray], titles: list[str], output_path: Path) -> None:
    fig, axes = plt.subplots(1, len(images), figsize=(2.5 * len(images), 2.8))
    for ax, image, title in zip(axes, images, titles):
        ax.imshow(image, vmin=0.0, vmax=1.0)
        # ax.set_title(title, fontsize=10)
        ax.axis("off")
    fig.tight_layout()
    fig.savefig(output_path, dpi=200, bbox_inches="tight")
    plt.close(fig)


sample = sample_image(image_total)

random_rotate = _make_random_rotate(num_channels=3)
random_zoom = _make_random_zoom(num_channels=3)
random_rotate_zoom = _make_random_rotate_zoom(num_channels=3)

save_row(
    [
        sample,
        apply_transform(sample, random_rotate, jax.random.PRNGKey(0)),
        apply_transform(sample, random_zoom, jax.random.PRNGKey(0)),
        apply_transform(sample, random_rotate_zoom, jax.random.PRNGKey(0)),
    ],
    [
        "original",
        "rotate",
        "zoom",
        "rotate+zoom",
    ],
    OUTPUT_DIR / "cifar_visualisation.pdf",
)
```

```python
import matplotlib.pyplot as plt
from matplotlib.legend_handler import HandlerTuple
import matplotlib as mpl
from utils.plotting import CISeries, plot_CI_trend_fn, plot_CI_freq_fn
import numpy as np

alpha = 0.05
d = 21
k = 50
class_str = 'SIGNAL-BACKGROUND'

folder_n=f'logs/WB180higgs_n50-2000_d{d}_k{k}_alpha{alpha:.2f}_testparam{class_str}_seed0_sim1000'
n_series = [
    CISeries(
        label=f'{method_label}',
        paths=[f'{folder_n}/_n{n}_k{k}.npz' for n in range(50,2005,50)],
        varfield='n',
        alpha=alpha,
        CIfield=CIfield,
        color=color,
        linestyle="-o",
        hist_linestyle="-",
        linewidth=1,
        markersize=5,
    )
    for method_label, CIfield, color in zip(
        ['Wild bootstrap', 'WB + Rotate'], 
        ['WB_compose_id_cis', 'WB_compose_random_rotate_cis'],
        ['tab:blue', 'tab:red']
    )
]

n = 2000
folder_k=f'logs/WB180higgs_n{n}_d{d}_k5-99_alpha{alpha:.2f}_testparam{class_str}_seed0_sim1000'
k_series = [
    CISeries(
        label=f'{method_label}',
        paths=[f'{folder_k}/_n{n}_k{k}.npz' for k in range(5,101,2)],
        varfield='k',
        alpha=alpha,
        CIfield=CIfield,
        color=color,
        linestyle="-o",
        hist_linestyle="-",
        linewidth=1,
        markersize=5,
    )
    for method_label, CIfield, color in zip(
        ['Wild bootstrap', 'WB + Rotate'], 
        ['WB_compose_id_cis', 'WB_compose_random_rotate_cis'],
        ['tab:blue', 'tab:red']
    )
]



fig, axs = plt.subplots(2, 2, figsize=(12, 6))

plot_CI_trend_fn(
    fn=lambda l,m,u: 1 - (l <= 0. and 0. <= u),
    series=n_series,
    xlim=[50,2000],
    ylim=[0.,1.05],
    fontsize=12,
    ax=axs[0][0],
)
axs[0][0].xaxis.set(visible=False)

plot_CI_trend_fn(
    fn=lambda l,m,u: u-l,
    series=n_series,
    xlim=[50,2000],
    ylim=[0.4,0.85],
    fontsize=12,
    ax=axs[1][0],
)

plot_CI_trend_fn(
    fn=lambda l,m,u: 1 - (l <= 0. and 0. <= u),
    series=k_series,
    xlim=[5,95],
    ylim=[0.,1.05],
    fontsize=12,
    ax=axs[0][1],
)
axs[0][1].xaxis.set(visible=False)
axs[0][1].yaxis.set(visible=False)

plot_CI_trend_fn(
    fn=lambda l,m,u: u-l,
    series=k_series,
    xlim=[5,95],
    ylim=[0.4,0.85],
    fontsize=12,
    ax=axs[1][1],
)
axs[1][1].yaxis.set(visible=False)

legend_by_label = {}
handles, labels = axs[0][0].get_legend_handles_labels()
for handle, label in zip(handles, labels):
    legend_by_label.setdefault(label, []).append(handle)

legend_handles = []
legend_labels = []
for legend_label, legend_group in legend_by_label.items():
    if len(legend_group) == 1:
        legend_handles.append(legend_group[0])
    else:
        legend_handles.append(tuple(legend_group))
    legend_labels.append(legend_label)

fig.tight_layout()
fig.subplots_adjust(wspace=0.2)
pos0 = axs[0][0].get_position()
shrink = pos0.height * 0.25
axs[0][0].set_position([pos0.x0, pos0.y0, pos0.width, pos0.height - shrink])
pos1 = axs[0][1].get_position()
shrink = pos1.height * 0.25
axs[0][1].set_position([pos1.x0, pos1.y0, pos1.width, pos1.height - shrink])


pos10 = axs[1][0].get_position()
shrink = pos10.height * 0.25
axs[1][0].set_position([pos10.x0, pos10.y0 + shrink, pos10.width, pos10.height - shrink])
pos11 = axs[1][1].get_position()
shrink = pos11.height * 0.25
axs[1][1].set_position([pos11.x0, pos11.y0 + shrink, pos11.width, pos11.height - shrink])



for ax_row in axs:
    for ax in ax_row:
        ax.tick_params(axis='both', labelsize=14)

fig.legend(
    legend_handles,
    legend_labels,
    loc='upper left',
    bbox_to_anchor=(pos0.x0+0.22, pos0.y0 + pos0.height + 0.07),
    bbox_transform=fig.transFigure,
    handlelength=4.0,
    handler_map={tuple: HandlerTuple(ndivide=None)},
    ncol=5,
    fontsize=15.5,
)

fig.savefig('figs/wb180higgs_overview_SIGNAL-BACKGROUND_vary_nk.pdf', dpi=100, bbox_inches='tight')
```
